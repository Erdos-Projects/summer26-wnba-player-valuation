# Model Testing

This notebook evaluates the final tuned model selected in **`model_tuning fixed.ipynb`**. The model is refit on the full 2021-2024 training set and tested once on the untouched 2025 holdout set.

### Objectives:

1. Load processed train/test files.
2. Load the best tuned estimator from `artifacts/best_model.joblib`.
3. Refit the model on all 2021-2024 training rows.
4. Predict 2025 salaries.
5. Save RMSE, MAE, MAPE, R², and player-level prediction errors.

```python
from pathlib import Path

import joblib
import numpy as np
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error, r2_score, root_mean_squared_error
```

```python
repo = Path.cwd().parent
# If the notebook is running from /notebooks, move one level up to the project root
if not (repo / "data").exists():
    repo = repo.parent

processed_dir = repo / "data" / "processed_salary_fix"
result_dir = repo / "results" / "Fixed"
art_dir = repo / "artifacts" / "Fixed"

x_train = pd.read_csv(processed_dir / "X_train_processed.csv")
y_train = pd.read_csv(processed_dir / "y_train.csv")["salary"]
x_test = pd.read_csv(processed_dir / "X_test_2025_processed.csv")
y_test = pd.read_csv(processed_dir / "y_test_2025.csv")["salary"]
lookup_test = pd.read_csv(processed_dir / "player_lookup_test_2025.csv")

model = joblib.load(art_dir / "best_model_fixed.joblib")
model
```

```python
model.fit(x_train, y_train)
pred = model.predict(x_test)

metrics = {
    "model": type(model).__name__,
    "train_years": "2021-2024",
    "test_year": 2025,
    "n_train": len(x_train),
    "n_test": len(x_test),
    "rmse": root_mean_squared_error(y_test, pred),
    "mae": mean_absolute_error(y_test, pred),
    "mape": mean_absolute_percentage_error(y_test, pred),
    "r2": r2_score(y_test, pred),
}

test_df = pd.DataFrame([metrics])
display(test_df)
```

## Player-Level Prediction Output

After evaluating the tuned model on the 2025 test set, save player-level predictions to `results/tuned_predictions.csv`.

This file combines each player's identifying information with the model output:

- `actual`: the player's observed 2025 salary
- `predicted`: the salary predicted by the tuned model
- `residual`: predicted salary minus actual salary
- `abs_error`: absolute prediction error

The `residual` column helps identify direction:

- positive residual means the model predicted a higher salary than the player actually earned
- negative residual means the model predicted a lower salary than the player actually earned

Sorting by `abs_error` shows the players with the largest prediction mistakes. This is useful as a diagnostic step because large errors may reveal contract types, roster situations, or player profiles that are not well captured by the current features.

```python
pred_df = lookup_test.copy()
pred_df["actual"] = y_test.values
pred_df["predicted"] = pred
pred_df["residual"] = pred_df["predicted"] - pred_df["actual"]
pred_df["abs_error"] = pred_df["residual"].abs()

pred_cols = ["player", "team", "year", "group", "actual", "predicted", "residual", "abs_error"]
pred_df = pred_df[pred_cols]
pred_df.to_csv(result_dir / "tuned_predictions_fixed.csv", index=False)
display(pred_df.sort_values("abs_error", ascending=False).head(10))
```
