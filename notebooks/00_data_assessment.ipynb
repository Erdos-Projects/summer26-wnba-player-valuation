```python
# Import pandas, and use pandas to anlayze tabular data
# Import path from pathlib and use 'path' to work with file paths
import pandas as pd
from pathlib import Path

ROOT = Path.cwd()

ROOT = ROOT.parent

# Define the location of the raw data folder.
# "../data/raw" means: starting from the current notebook folder, go up one level,then enter the data/raw folder.
RAW_DIR = ROOT / "data" / "raw"

# Create a list of raw data files that we want to inspect.
files = [
    "salary_2025.csv",
    "2025_advanced.csv",
    "2025_per_game.csv",
    "2025_totals.csv",
    "2025_advanced-team.csv",
    "2025_wnba_standings.csv",
]

# Loop through each file name in the files list.
for file in files:
    path = RAW_DIR / file
    df = pd.read_csv(path)
    print("=" * 80)
    print(file)
    print("shape:", df.shape)
    print("columns:", list(df.columns))
    print(df.head())
```

```python
import pandas as pd
from pathlib import Path

RAW_DIR = Path("../data/raw")

files = [
    "salary_2025.csv",
    "2025_advanced.csv",
    "2025_per_game.csv",
    "2025_totals.csv",
    "2025_advanced-team.csv",
    "2025_wnba_standings.csv",
]

summary = []

for file in files:
    path = RAW_DIR / file
    df = pd.read_csv(path)

    summary.append({
        "file_name": file,
        "num_rows": df.shape[0],
        "num_columns": df.shape[1],
        "columns": ", ".join(df.columns)
    })

summary_df = pd.DataFrame(summary)
summary_df
```
